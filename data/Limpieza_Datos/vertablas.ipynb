## Abrir Tablas

```python
df_central = pd.read_csv("../ttransporte_eod2017/conjunto_de_datos/ttransporte.csv")
df_tviaje = pd.read_csv("../tviaje_eod2017/conjunto_de_datos/tviaje.csv")
```

Buscar si hay ids de `id_via` que no están en ambos

```python
ids_central = set(df_central["id_via"])
ids_tviaje  = set(df_tviaje["id_via"])

missing_in_tviaje = ids_central - ids_tviaje

missing_in_central = ids_tviaje - ids_central

print("Missing in df_tviaje:", missing_in_tviaje)
print("Missing in df_central:", missing_in_central)
```

```python
df_central["id_via"].nunique()
```

```python
df_tviaje["id_via"].nunique()
```

Todos los ids están presentes; por lo que se juntarán ambas tablas mediante un left join

```python
df_merged = df_central.merge(
    df_tviaje[["id_via", "p5_6", "p5_13"]],
    on="id_via",
    how="left"
)
df_merged
```

Ahora checamos nans en df_merged en las columnas agregadas

```python
df_merged[["p5_13", "p5_6"]].isna().sum()
```

Se hizo correctamente el merge.
